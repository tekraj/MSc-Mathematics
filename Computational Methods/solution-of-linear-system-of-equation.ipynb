## Jacobi Method

```python
import numpy as np

A = np.array([
    [10, 1, 2],
    [2, 10, 1],
    [1, 2, 10]
])
b = np.array([13.0, 13.0, 13.0])

xk = np.array([0.0, 0.0, 0.0]) 
xk1 = np.array([0.0, 0.0, 0.0])

tol = 10E-2
max_iteration = 10000

for itr in range(0, max_iteration):
    xk = xk1.copy()
    for i in range(0, 3):
        row_sum = sum([xk1[j] * A[i, j] for j in range(0, 3) if i != j])
        xk1[i] = (1/A[i, i]) * (b[i] - row_sum)
    
    error = np.max(np.abs(xk1 - xk))
    
    print(f"Iteration {itr}, Error: {error:.4f}, x: {xk1}")
    
    if error < tol and itr > 0: 
        break

print("\nFinal result:", xk1)
```

```python
import numpy as np

# System Coefficients
A = np.array([
    [10.0, 1.0, 2.0],
    [2.0, 10.0, 1.0],
    [1.0, 2.0, 10.0]
])
b = np.array([13.0, 13.0, 13.0])

# Settings
omega = 1.033  # Relaxation factor (w > 1 for acceleration)
tol = 1e-4
max_iteration = 100
xk = np.array([0.0, 0.0, 0.0])

print(f"{'Iteration':<10} | {'x1':<10} | {'x2':<10} | {'x3':<10} | {'Error':<10}")
print("-" * 65)

for itr in range(max_iteration):
    # 1. Take a snapshot of the starting values to calculate error later
    xk_old = xk.copy()
    
    # 2. Update each variable one-by-one (The Row Loop)
    for i in range(len(b)):
        # Calculate the sum of all OTHER terms in the row
        # This uses the most CURRENT values in 'xk' (Immediate Update)
        row_sum = 0
        for j in range(len(b)):
            if i != j:
                row_sum += A[i, j] * xk[j]
        
        # Calculate the standard Gauss-Seidel value
        gs_value = (b[i] - row_sum) / A[i, i]
        
        # Apply Relaxation: Weighted average of OLD and NEW
        # x_new = (1 - w) * x_old + (w * gs_new)
        xk[i] = (1 - omega) * xk[i] + (omega * gs_value)

    # 3. Check how much the entire vector moved
    error = np.max(np.abs(xk - xk_old))
    
    print(f"{itr:<10} | {xk[0]:<10.4f} | {xk[1]:<10.4f} | {xk[2]:<10.4f} | {error:<10.6f}")

    if error < tol:
        print("\nConvergence reached!")
        break

print("\nFinal Solution Vector:", xk)
```
