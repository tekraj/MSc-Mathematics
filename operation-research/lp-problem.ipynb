```python
import matplotlib.pyplot as plt
import numpy as np

# Constraints:
# 1. 4x1 + 6x2 <= 360  =>  x2 <= (360 - 4x1) / 6
# 2. 3x1 <= 180        =>  x1 <= 60
# 3. 5x2 <= 200        =>  x2 <= 40
# x1, x2 >= 0

x1_vals = np.linspace(0, 100, 400)

# Define lines
y1 = (360 - 4*x1_vals) / 6
y2 = np.full_like(x1_vals, 40)
x_const = 60

# Objective function: Z = 15x1 + 10x2
def objective(x1, x2):
    return 15*x1 + 10*x2

# Intersection points (vertices of the feasible region)
# 1. (0,0)
# 2. (60, 0) -> From x1 = 60 and x2 = 0
# 3. (60, 20) -> From x1 = 60 and 4(60) + 6x2 = 360 => 240 + 6x2 = 360 => 6x2 = 120 => x2 = 20
# 4. (30, 40) -> From x2 = 40 and 4x1 + 6(40) = 360 => 4x1 + 240 = 360 => 4x1 = 120 => x1 = 30
# 5. (0, 40) -> From x1 = 0 and x2 = 40

vertices = [(0, 0), (60, 0), (60, 20), (30, 40), (0, 40)]
results = {v: objective(v[0], v[1]) for v in vertices}

# Plotting
plt.figure(figsize=(10, 8))
plt.plot(x1_vals, y1, label=r'$4x_1 + 6x_2 \leq 360$', color='blue')
plt.axvline(x=x_const, label=r'$3x_1 \leq 180$', color='green')
plt.axhline(y=40, label=r'$5x_2 \leq 200$', color='red')

# Fill feasible region
path_x = [0, 60, 60, 30, 0]
path_y = [0, 0, 20, 40, 40]
plt.fill(path_x, path_y, color='grey', alpha=0.3, label='Feasible Region')

# Label vertices
for v in vertices:
    plt.plot(v[0], v[1], 'ko')
    plt.text(v[0]+1, v[1]+1, f'({v[0]}, {v[1]})\nZ={results[v]}')

plt.xlim(0, 100)
plt.ylim(0, 80)
plt.xlabel('$x_1$')
plt.ylabel('$x_2$')
plt.title('Graphical Solution for Linear Programming')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)
plt.savefig('solution_plot.png')
plt.show()

print(results)
```

### LP Problem

```python
import numpy as np
import matplotlib.pyplot as plt


def lp_equation(a,b,c):
    x = np.linspace(0, 10, 100)
    y = (c - a*x) / b
    plt.plot(x, y, label=f'{a}x + {b}y = {c}')
    plt.xlabel('x')
    plt.ylabel('y')
    plt.title('Graph of the Linear Equation')
    plt.axhline(0, color='black', lw=0.5)
    plt.axvline(0, color='black', lw=0.5)
    plt.grid()
    plt.legend()
    plt.show()


lp_equation(1, 10, 20)
lp_equation(1, 10, 30)
```
