```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set seed for reproducibility
np.random.seed(42)

# Generate Synthetic Data (N=1000)
n = 1000
data = {
    'age': np.random.randint(18, 75, n),
    'workclass': np.random.choice(['Private', 'Self-Emp', 'Government', 'Unknown'], n),
    'education_num': np.random.randint(1, 17, n),
    'marital_status': np.random.choice(['Married', 'Single', 'Divorced'], n),
    'occupation': np.random.choice(['Tech', 'Sales', 'Service', 'Professional'], n),
    'hours_per_week': np.random.randint(10, 70, n),
    'capital_gain': np.random.choice([0, 5000, 10000, 25000], n, p=[0.8, 0.1, 0.07, 0.03]),
    'income': np.random.choice(['<=50K', '>50K'], n)
}

df = pd.DataFrame(data)

# Inject some correlation logic: High Education + High Age = More likely >50K
mask = (df['education_num'] > 12) & (df['age'] > 35)
df.loc[mask, 'income'] = np.random.choice(['<=50K', '>50K'], len(df[mask]), p=[0.2, 0.8])

df.to_csv("adlt_data.csv", index=False)
```
